```python
import sqlite3

conn = sqlite3.connect("a.db")
cursor = conn.cursor()

cursor.execute('''
    CREATE TABLE IF NOT EXISTS abs(
    ID INTEGER, Name TEXT, email TEXT 
    )
''')

num = int(input("enteries"))
for _ in range(num):
    id2 = int(input())
    name = input("")
    email = input("")

    cursor.execute("INSERT INTO abs VALUES(?, ?, ?)", (id2, name, email))
cursor.execute("SELECT * FROM abs")
print(cursor.fetchall())
conn.commit()
conn.close()
```
