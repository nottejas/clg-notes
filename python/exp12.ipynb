```python
import sqlite3
conn = sqlite3.connect('student.db')

cursor = conn.cursor()

cursor.execute('''CREATE TABLE IF NOT EXISTS students (

                    name TEXT,

                    roll_number TEXT UNIQUE,

                    sap_id TEXT UNIQUE,

                    program TEXT,

                    semester INTEGER,

                    marks INTEGER,

                    result TEXT CHECK(result IN('Pass','Fail'))

                )''')

conn.commit()
 
```

```python
cursor.execute("INSERT INTO students (name, roll_number, sap_id, program, semester, marks, result) VALUES (?, ?, ?, ?, ?, ?, ?)",
               ('John Doe', '12345', 'S123456', 'Computer Science', 2, 85, 'Pass'))
cursor.execute("INSERT INTO students (name, roll_number, sap_id, program, semester, marks, result) VALUES (?, ?, ?, ?, ?, ?, ?)",
               ('Jane Smith', '67890', 'S654321', 'Electrical Engineering', 4, 75, 'Pass'))
conn.commit()
```

```python
cursor.execute("SELECT * FROM students")
students = cursor.fetchall()
students
```
